```python
# 필수 1 코드를 작성하세요.
import math
import numpy as np
import pandas as pd
from scipy import stats

# statsmodels가 있으면 그대로 쓰고, 없으면 scipy로 같은 계산을 합니다.
try:
    from statsmodels.stats.power import TTestIndPower
    analysis = TTestIndPower()
    use_sm = True
except ModuleNotFoundError:
    analysis = None
    use_sm = False

def get_power(effect_size, n, alpha):
    """그룹당 표본 수가 n일 때의 검정력(양측검정)"""
    if use_sm:
        return analysis.power(effect_size=effect_size, nobs1=n, alpha=alpha,
                              ratio=1.0, alternative="two-sided")
    dof = 2 * n - 2                         # 자유도
    ncp = effect_size * math.sqrt(n / 2)    # 효과 크기가 만드는 분포 이동량
    t_crit = stats.t.ppf(1 - alpha / 2, dof)
    return stats.nct.sf(t_crit, dof, ncp) + stats.nct.cdf(-t_crit, dof, ncp)

def get_n(effect_size, target_power, alpha):
    """목표 검정력을 채우는 그룹당 표본 수"""
    if use_sm:
        return analysis.solve_power(effect_size=effect_size, power=target_power,
                                    alpha=alpha, ratio=1.0, alternative="two-sided")
    n = 2
    while get_power(effect_size, n, alpha) < target_power:
        n += 1
    return n

# 조건 설정
effect_size = 0.3
alpha = 0.05
target_power = 0.8
sample_sizes = [30, 100, 300]

# 2 & 3. 표본 크기별 검정력  ← 기존 print에 제2종 오류 확률만 추가
print("=== 표본 크기별 검정력 (양측검정) ===")
power_results = {}
for n in sample_sizes:
    power = get_power(effect_size, n, alpha)
    power_results[n] = power
    print(f"그룹당 표본 수 {n:>3}명 : 검정력 = {power:.3f} / 놓칠 확률(제2종 오류) = {1 - power:.3f}")

# 4 & 5. 목표 검정력 0.8에 필요한 표본 수  ← 참고 줄 한 개 추가
required_n = get_n(effect_size, target_power, alpha)
required_n_ceil = math.ceil(required_n)
print("\n=== 목표 검정력 0.8을 위한 표본 크기 ===")
print(f"계산된 필요 표본 수(소수) : {required_n:.4f}")
print(f"최종 필요 표본 수(올림)   : 그룹당 {required_n_ceil}명")
print(f"참고) {required_n_ceil - 1}명 검정력 {get_power(effect_size, required_n_ceil - 1, alpha):.3f} (0.8 미달) / "
      f"{required_n_ceil}명 검정력 {get_power(effect_size, required_n_ceil, alpha):.3f} (충족)")

# 6. 목표 검정력 충족 여부
print("\n=== 목표 검정력(0.8) 충족 여부 ===")
for n in sample_sizes:
    status = "충족" if power_results[n] >= target_power else "미충족"
    print(f"그룹당 {n:>3}명 : {status} (검정력 {power_results[n]:.3f})")
```

### 문제 2-1. 중앙 냉방시설 유무에 따른 판매가격 차이 분석

#### 문제 설명

한 부동산 회사는 중앙 냉방시설이 있는 주택과 없는 주택의 평균 판매가격 차이를 분석하려고 합니다. 회사는 두 집단의 평균 판매가격 차이가 **100,000달러 이상**이어야 냉방시설 설치 지원 사업을 검토할 실무적 가치가 있다고 정했습니다.

#### 요구사항

   - `p-value ≤ 0.05`: 통계적으로 유의함
   - `|Cohen’s d| ≥ 0.8`: 큰 효과
   - `|평균 차이| ≥ 100000`: 회사 기준에서 실질적으로 유의함
8. 세 결과를 종합하여 사업 검토 여부에 대한 근거를 작성하세요.

#### Cohen’s d 공식

두 집단의 평균 차이를 합동 표준편차로 나누어 계산합니다.

$$
d = \frac{\bar{x}_1 - \bar{x}_2}{s_p}
$$

$$
s_p = \sqrt{\frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}}
$$

- $\bar{x}_1$, $\bar{x}_2$: 각 집단의 평균
- $s_1$, $s_2$: 각 집단의 표본 표준편차
- $n_1$, $n_2$: 각 집단의 표본 수
- $s_p$: 합동 표준편차

#### 제출 결과

- 집단별 표본 수와 평균
- 독립표본 t검정 결과
- 평균 차이와 Cohen’s d
- 통계적·효과크기·회사 기준 판단
- 최종 의사결정 근거
- Q1~Q5 답변

```python
# 회사 기준으로는 10만달러로 기준이었는데 기준을 충족하지 못하였음. 
# 최종판안 : 일부 통계적 근거는 있지만 회사의 모든 기준을 충족하지는 않습니다.

# 필수 2 코드를 작성하세요.
import math
import pandas as pd
from scipy import stats

# 1 & 2. 집단 나누기
air_yes = df[df["CentralAir"] == "Y"]["SalePrice"]
air_no = df[df["CentralAir"] == "N"]["SalePrice"]

# 3. 표본 수와 평균
print("=== 집단별 표본 수와 평균 ===")
print(f"냉방시설 있음(Y) : {len(air_yes)}채, 평균 {air_yes.mean():,.1f}달러")
print(f"냉방시설 없음(N) : {len(air_no)}채, 평균 {air_no.mean():,.1f}달러")

# 4. 독립표본 t검정 (양측)
t_stat, p_value = stats.ttest_ind(air_yes, air_no, equal_var=True)

# 5. Cohen's d 계산 함수
def cohen_d(group1, group2):
    n1, n2 = len(group1), len(group2)
    s1, s2 = group1.std(ddof=1), group2.std(ddof=1)
    sp = math.sqrt(((n1 - 1) * s1 ** 2 + (n2 - 1) * s2 ** 2) / (n1 + n2 - 2))
    return (group1.mean() - group2.mean()) / sp

mean_diff = air_yes.mean() - air_no.mean()
d = cohen_d(air_yes, air_no)

# 6. 결과 출력
print("\n=== 검정 결과 ===")
print(f"t통계량   : {t_stat:.3f}")
print(f"p-value   : {p_value:.3e}")
print(f"평균 차이 : {mean_diff:,.1f}달러")
print(f"Cohen's d : {d:.3f}")

# 7. 세 가지 기준 판단
alpha = 0.05
d_cut = 0.8
money_cut = 100000

stat_sig = p_value <= alpha
big_effect = abs(d) >= d_cut
practical = abs(mean_diff) >= money_cut

print("\n=== 기준별 판단 ===")
print(f"통계적 유의성 (p <= {alpha})        : {'충족' if stat_sig else '미충족'}")
print(f"큰 효과 (|d| >= {d_cut})             : {'충족' if big_effect else '미충족'}")
print(f"실질적 유의성 (|차이| >= {money_cut:,}) : {'충족' if practical else '미충족'}")

# 8. 종합 의사결정 근거
print("\n=== 최종 의사결정 근거 ===")
if stat_sig and big_effect and practical:
    print("세 기준을 모두 충족하므로 냉방시설 설치 지원 사업을 검토할 근거가 충분합니다.")
else:
    print(f"통계적으로는 유의하고(p={p_value:.3e}) 효과 크기도 크지만(d={d:.3f}),")
    print(f"평균 차이 {mean_diff:,.1f}달러가 회사 기준 {money_cut:,}달러에 "
          f"{money_cut - abs(mean_diff):,.1f}달러 부족합니다.")
    print("따라서 통계적 근거는 있으나 회사의 실무 기준은 충족하지 못해 사업 검토는 보류가 타당합니다.")
```

### 문제 3-1. 주방 품질이 `Gd`인 집단과 `TA`인 집단 비교

#### 문제 설명

부동산 중개회사는 주방 품질이 `Gd`(Good)인 주택과 `TA`(Typical/Average)인 주택의 평균 판매가격을 비교하려고 합니다. 회사는 평균 판매가격 차이가 **50,000달러 이상**이면 마케팅에서 주방 품질의 차이를 강조할 실무적 가치가 있다고 판단합니다.

> 필수 2에서 학습한 검정과 효과 크기 계산 절차를 새로운 두 집단에 적용하는 과제입니다.

#### 요구사항

1. 주방 품질이 `Gd`인 집단과 `TA`인 집단의 `SalePrice`를 각각 준비하세요.
2. 두 집단의 표본 수와 평균 판매가격을 출력하세요.
3. 두 집단의 평균 차이에 대한 양측 독립표본 t검정을 수행하세요.
4. Cohen’s d를 계산하세요.
5. 평균 차이, p-value, Cohen’s d를 출력하세요.
6. 다음 기준에 따라 각각 판단하세요.
   - 통계적 유의성: `p-value ≤ 0.05`
   - 큰 효과: `|Cohen’s d| ≥ 0.8`
   - 실무적 유의성: `|평균 차이| ≥ 50000`
7. 세 가지 판단을 종합하여 주방 품질을 마케팅에서 강조할 근거가 있는지 결론을 작성하세요.

#### 해석 질문

**Q1.** 두 집단의 판매가격 차이는 통계적으로 유의한가요?  
**Q2.** Cohen’s d를 기준으로 효과 크기는 어느 정도인가요?  
**Q3.** 회사가 정한 50,000달러 기준을 충족하나요?  
**Q4.** 최종적으로 주방 품질 차이를 마케팅에서 강조할 근거가 있다고 볼 수 있나요?

#### 제출 결과

- 집단별 표본 수와 평균
- t통계량과 p-value
- 평균 차이와 Cohen’s d
- 통계적·실질적 유의성 판단
- 최종 결론
- Q1~Q4 답변

```python
# 과제 코드를 작성하세요.

import math
import pandas as pd
from scipy import stats

def cohen_d(group1, group2):
    n1, n2 = len(group1), len(group2)
    s1, s2 = group1.std(ddof=1), group2.std(ddof=1)
    sp = math.sqrt(((n1 - 1) * s1 ** 2 + (n2 - 1) * s2 ** 2) / (n1 + n2 - 2))
    return (group1.mean() - group2.mean()) / sp

# 1. 두 집단 준비
gd_price = df[df["KitchenQual"] == "Gd"]["SalePrice"]
ta_price = df[df["KitchenQual"] == "TA"]["SalePrice"]

# 2. 표본 수와 평균
print("=== 집단별 표본 수와 평균 ===")
print(f"주방 품질 Gd : {len(gd_price)}채, 평균 {gd_price.mean():,.1f}달러")
print(f"주방 품질 TA : {len(ta_price)}채, 평균 {ta_price.mean():,.1f}달러")

# 3. 독립표본 t검정 (양측)
t_stat, p_value = stats.ttest_ind(gd_price, ta_price, equal_var=True)

# 4 & 5. 평균 차이, Cohen's d 출력
mean_diff = gd_price.mean() - ta_price.mean()
d = cohen_d(gd_price, ta_price)

print("\n=== 검정 결과 ===")
print(f"t통계량   : {t_stat:.3f}")
print(f"p-value   : {p_value:.3e}")
print(f"평균 차이 : {mean_diff:,.1f}달러")
print(f"Cohen's d : {d:.3f}")

# 6. 기준별 판단
alpha = 0.05
d_cut = 0.8
money_cut = 50000

stat_sig = p_value <= alpha
big_effect = abs(d) >= d_cut
practical = abs(mean_diff) >= money_cut

print("\n=== 기준별 판단 ===")
print(f"통계적 유의성 (p <= {alpha})       : {'충족' if stat_sig else '미충족'}")
print(f"큰 효과 (|d| >= {d_cut})            : {'충족' if big_effect else '미충족'}")
print(f"실질적 유의성 (|차이| >= {money_cut:,}) : {'충족' if practical else '미충족'}")

# 7. 최종 결론
print("\n=== 최종 결론 ===")
if stat_sig and big_effect and practical:
    print(f"세 기준을 모두 충족합니다. 주방 품질이 Gd인 주택이 TA보다 평균 "
          f"{mean_diff:,.1f}달러 비싸고(p={p_value:.3e}, d={d:.3f}),")
    print(f"회사 기준 {money_cut:,}달러도 넘으므로 주방 품질 차이를 마케팅에서 강조할 근거가 있습니다.")
else:
    print("세 기준 중 일부를 충족하지 못하므로 마케팅에서 강조하기에는 근거가 부족합니다.")
```
