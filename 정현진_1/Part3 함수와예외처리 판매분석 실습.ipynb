# 문제 1. 판매금액 구간 분류 함수 만들기

## 함수

```python
classify_sales_by_amount(sale_ids, sale_amounts)
```

## 요구사항

1. 두 입력값이 리스트가 아니면 `TypeError`를 발생시킵니다.
2. 두 리스트의 길이가 다르면 `ValueError`를 발생시킵니다.
3. 데이터가 비어 있으면 `ValueError`를 발생시킵니다.
4. 고가·중가·일반 거래의 `SaleID`를 딕셔너리로 반환합니다.
5. 함수를 호출하고 결과를 출력합니다.

````python
# 문제 1 코드를 작성하세요.
def classify_sales_by_amount(sale_amounts, sale_ids):
    if not isinstance (sale_ids,list) or not isinstance (sale_amounts,list):
        raise TypeError("두 입력값은 리스트이어야 합니다.") #리스트 아니면 TypeError
    if len(sale_ids) != len(sale_amounts):
        raise ValueError("두 리스트의 길이는 같아야 합니다.")
    if len(sale_ids) == 0:
        raise ValueError("데이터가 존재해야 합니다.")

    
    high_saleids=[]
    mid_saleids=[]
    nor_saleids=[]

    for amount, sale_id in zip(sale_amounts, sale_ids):
        if amount >= 70000000:
            high_saleids.append(sale_id)
        elif amount >= 40000000 :
            mid_saleids.append(sale_id)
        else:
            nor_saleids.append(sale_id)
    return {"고가": high_saleids,"중가": mid_saleids,"저가": nor_saleids}

result = classify_sales_by_amount(sale_amounts, sale_ids)
print("고가:", len(result["고가"]))
print("중가:", len(result["중가"]))
print("저가:", len(result["저가"]))

# high_count = 0
# mid_count = 0
# nor_count = 0

# high_saleids=[]
# mid_saleids=[]
# nor_saleids=[]

# for amount in range(len(sale_amounts)):
#         if sale_amounts[amount] >= 70000000:
#             high_count += 1
#             high_saleids.append(sale_ids[amount])
            
#         elif sale_amounts[amount] >= 40000000 and amount < 70000000:
#             mid_count += 1
#             mid_saleids.append(sale_ids[amount])
#         else:
#             nor_count += 1
#             nor_saleids.append(sale_ids[amount])

# print("고가거래:", high_count, "건", high_saleids[1:6])
# print("중가거래:", mid_count, "건", mid_saleids[1:6])
# print("일반거래:", nor_count, "건", nor_saleids[1:6])

# print(total_counts == high_count + mid_count + nor_count)

````

# 문제 2. 배송 확인 거래 검색 함수 만들기

## 함수

```python
find_delivery_check_sales(
    sale_ids,
    delivery_days,
    stock_statuses
)
```

## 요구사항

1. 세 입력값이 리스트가 아니면 `TypeError`를 발생시킵니다.
2. 세 리스트의 길이가 다르면 `ValueError`를 발생시킵니다.
3. 배송일 20일 이상이면서 출고완료가 아닌 거래의 `SaleID`를 반환합니다.
4. 정상 입력과 잘못된 입력을 각각 테스트합니다.
5. `TypeError`와 `ValueError`를 별도의 `except`에서 처리합니다.

````python
# 문제 2 코드를 작성하세요.
def find_delivery_check_sales(sale_ids, delivery_days, stock_statuses):
    if not isinstance(sale_ids, list) or not isinstance(delivery_days, list) or not isinstance(stock_statuses, list):
        raise TypeError("세 입력값은 리스트이어야 합니다.")
    if len(sale_ids) != len(delivery_days) or len(sale_ids) != len(stock_statuses):
        raise ValueError("세 리스트의 길이는 같아야 합니다.")

    check = []

    for ids, days, status in zip(sale_ids, delivery_days, stock_statuses):
        if days >= 20 and status != "출고완료":
            check.append(ids)

    return check
print(find_delivery_check_sales(sale_ids, delivery_days, stock_statuses))

# 테스트
try:
    result = find_delivery_check_sales(sale_ids, delivery_days, stock_statuses)
    print("출고 확인 대상:", result)

except TypeError as count:
    print("TypeError:", count)

except ValueError as count:
    print("ValueError:", count)
````
