#### 요구사항

   - 전체 구독 수

#### 해석 질문

**Q1.** 요금제별 이탈률을 비교하는 것이 전체 이탈률만 확인하는 것보다 행동 가능성이 높은 이유는 무엇인가요
-> 전체 이탈률은 서비스 전체의 현상을 보여주고, 요금제별 이탈률은 어느 요금제를 먼저 점검해야하는 지를 판단한다.
-> 다음 회의가 열린다면 담당 팀이 개선 대상을 명확하게 설정하여 개선점을 진행할 수 있다.

**Q2.** 현재 유료 구독 MRR이 가장 큰 요금제는 무엇인가요?  
-> Enterprise이고 현재 유로 구독료는 $7,546,876이다. 

**Q3.** 이탈률이 가장 높은 요금제는 무엇인가요?
-> Enterprise이고 이탈률은 9.98%이다.

**Q4.** 위 두 결과를 함께 보면 어떤 요금제를 우선적으로 점검할 수 있으며, 그 이유는 무엇인가요?
-> Enterprise를 우선적으로 점검한다.
-> 현재 MRR 기여도가 가장 크면서 이탈률이 가장 높아 
이 그룹의 이탈률을 줄이는 것이 다음달 매출의 방어 효과가 가장 크다.

#### 제출 결과

- `plan_metrics` 집계 코드와 결과
- MRR 및 이탈률 기준 요금제 비교
- 우선 점검 대상과 개선 방향
- Q1~Q4 답변

```python
# 필수 2 코드를 작성하세요.
subscriptions["is_active_paid"] = (
    subscriptions["end_date"].isna()
    & (subscriptions["is_trial"] == False))

plan_metrics = subscriptions.groupby("plan_tier").agg(
    total_subscriptions=("subscription_id", "nunique"),
    churned_subscriptions=("churn_flag", "sum"),
    churn_rate=("churn_flag", "mean"),
    active_paid_subscriptions=("is_active_paid", "sum"))

active_mrr_by_plan = (
    subscriptions.loc[subscriptions["is_active_paid"]]
    .groupby("plan_tier")["mrr_amount"]
    .sum()
)
plan_metrics["active_mrr"] = active_mrr_by_plan

print(plan_metrics.to_string(formatters={
    "churn_rate": lambda x: f"{x:.2%}",
    "active_mrr": lambda x: f"${x:,.0f}"
}))

highest_mrr_plan = plan_metrics["active_mrr"].idxmax()
highest_churn_plan = plan_metrics["churn_rate"].idxmax()

print(f"현재 유료 구독 MRR이 가장 큰 요금제: {highest_mrr_plan}")
print(f"구독 이탈률이 가장 높은 요금제: {highest_churn_plan}")
print("우선 점검 대상: Enterprise")
print("개선 방향: Enterprise 이탈 고객의 이유와 사용 패턴을 확인합니다.")
```

```python
# 과제 1 코드를 작성하세요.
source_metrics = accounts.groupby("referral_source").agg(
    total_accounts=("account_id", "nunique"),
    churned_accounts=("churn_flag", "sum"),
    churn_rate=("churn_flag", "mean"))

overall_churn_rate = accounts["churn_flag"].mean()
source_metrics["gap_vs_overall"] = source_metrics["churn_rate"] - overall_churn_rate

source_metrics = source_metrics.sort_values("churn_rate", ascending=False)

print(source_metrics.to_string(formatters={
    "churn_rate": lambda x: f"{x:.2%}",
    "gap_vs_overall": lambda x: f"{x:+.2%}p"
}))

print(f"\n전체 고객사 이탈률: {overall_churn_rate:.2%}")

highest_churn_source = source_metrics["churn_rate"].idxmax()
lowest_churn_source = source_metrics["churn_rate"].idxmin()

highest_churn_rate = source_metrics.loc[highest_churn_source, "churn_rate"]
lowest_churn_rate = source_metrics.loc[lowest_churn_source, "churn_rate"]
highest_total = source_metrics.loc[highest_churn_source, "total_accounts"]
highest_churned = source_metrics.loc[highest_churn_source, "churned_accounts"]

print(f"\n이탈률이 가장 높은 유입 경로: {highest_churn_source} ({highest_churn_rate:.2%})")
print(f"- 고객사 수: {highest_total:,}개 / 이탈 고객사 수: {highest_churned:,}개")
print(f"- 전체 이탈률 대비: {highest_churn_rate - overall_churn_rate:+.2%}p")
print(f"이탈률이 가장 낮은 유입 경로: {lowest_churn_source} ({lowest_churn_rate:.2%})")
print(f"경로 간 이탈률 격차: {highest_churn_rate - lowest_churn_rate:.2%}p")

print(f"\n우선 점검 대상: {highest_churn_source}")
print(f"개선 방향: {highest_churn_source} 경로로 들어온 고객사의 가입 직후 온보딩 과정을 점검하여 초기 정착률을 높인다.")
```
