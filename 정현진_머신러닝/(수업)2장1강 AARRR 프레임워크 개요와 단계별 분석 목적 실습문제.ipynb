Ravenstack은 기업 고객에게 구독형 소프트웨어를 제공하는 B2B SaaS 서비스입니다.

Ravenstack의 AARRR 단계는 다음과 같이 연결할 수 있습니다.

| 단계 | Ravenstack에서의 의미 | 활용 데이터 예시 |
|---|---|---|
| Acquisition | 고객사가 유입되어 가입함 | 가입일, 유입 경로 |
| Activation | 구독 후 제품 기능을 처음 사용함 | 기능 사용 기록 |
| Retention | 제품 사용과 구독을 계속 유지함 | 반복 사용 기록, 이탈 여부 |
| Referral | 기존 고객이 다른 고객을 추천함 | 추천 전송·추천 가입 이벤트 |
| Revenue | 유료 구독에서 반복 매출이 발생함 | MRR, 요금제 |

> 제공된 데이터에는 고객 추천 행동을 직접 기록한 이벤트가 없습니다. `referral_source`는 고객의 유입 경로이며, 기존 고객이 추천 행동을 했다는 사실을 직접 보여주는 Referral 이벤트와는 구분해야 합니다.

## 필수 1. Ravenstack의 AARRR 지표 지도 만들기

### 문제 1-1. 데이터에서 확인 가능한 단계별 대표 지표 계산하기

#### 문제 설명

Ravenstack의 세 테이블을 이용하여 AARRR 단계별 대표 지표를 계산하고, 제공된 데이터만으로 직접 측정하기 어려운 단계도 확인하세요.

이번 문제에서는 다음 기준을 사용합니다.

- Acquisition: 가입한 고유 고객사 수
- Activation: 전체 구독 중 기능 사용 기록이 한 번 이상 있는 구독의 비율
- Retention: 전체 구독 중 현재 `churn_flag`가 `False`인 구독의 비율
- Referral: 제공된 데이터로 직접 계산할 수 없음
- Revenue: 종료일이 없고 체험 구독이 아닌 현재 활성 유료 구독의 MRR 합계

> 여기에서 Retention은 학습을 위한 간단한 대리 지표입니다. 특정 기간 후 다시 사용한 비율을 계산한 정식 리텐션과는 다릅니다.

#### 요구사항

1. 가입한 고유 고객사 수를 `acquired_accounts`로 계산하세요.
2. 기능 사용 기록이 있는 고유 구독 수를 전체 구독 수로 나누어 `activation_rate`를 계산하세요.
3. `churn_flag`가 `False`인 구독의 비율을 `non_churned_rate`로 계산하세요.
4. 종료일이 없고 체험 구독이 아닌 구독의 MRR 합계를 `current_paid_mrr`로 계산하세요.
5. 네 지표를 알아보기 쉬운 형식으로 출력하세요.

#### 해석 질문

**Q1.** 기능 사용 기록이 있는 구독 비율은 어느 AARRR 단계와 연결되나요?

**Q2.** `non_churned_rate`를 정식 리텐션율과 동일하게 보면 안 되는 이유는 무엇인가요?  
-> 비이탈 상태만 나타나고, 동일한 사용자 그룹이 가입 후 언제 사용했고 다시 사용했는 지를
-> 시간에 따라 추적하지 않아서 

**Q3.** `referral_source`만으로 Referral 행동을 직접 측정할 수 있나요?  
-> 어렵다. referral_sourc는 고객이 어떤 경로로 들어왔는지 보여주지만, 이 고객이 다른 고객을 추천해주거나 유도했다는 행동 자체는 보여주지 않는다.
-> 이벤트성 변수가 추가적으로 필요하다. 

**Q4.** MRR은 어느 AARRR 단계와 연결되나요?

#### 제출 결과

- AARRR 단계별 대표 지표 계산 코드와 결과
- 각 지표의 단계 및 분석 목적
- Referral 측정에 필요한 이벤트 제안
- Q1~Q4 답변

```python
# 필수 1 코드를 작성하세요.
acquired_accounts = accounts["account_id"].nunique()

total_subscriptions = subscriptions["subscription_id"].nunique()
used_subscriptions = feature_usage.loc[
    feature_usage["usage_count"] > 0,
    "subscription_id"].nunique()

activation_rate = used_subscriptions / total_subscriptions
non_churned_rate = (~subscriptions["churn_flag"]).mean()
active_paid_mask = (subscriptions["end_date"].isna() & (subscriptions["is_trial"] == False))
current_paid_mrr = subscriptions.loc[active_paid_mask, "mrr_amount"].sum()

print(f"Acquisition - 가입 고객사 수: {acquired_accounts:,}개")
print(f"Activation - 기능 사용 기록이 있는 구독 비율: {activation_rate:.2%}")
print(f"Retention 대리 지표 - 현재 비이탈 구독 비율: {non_churned_rate:.2%}")
print("Referral - 제공된 데이터로 직접 계산할 수 없음")
print(f"Revenue - 현재 활성 유료 구독 MRR: ${current_paid_mrr:,.0f}")
```

## 필수 2. 요금제별 Activation 지표 비교하기

이번 실습에서 제품 활성률은 다음과 같이 정의합니다.

> 12월 제품 활성 유료 구독 수 ÷ 12월 이용 가능 유료 구독 수

#### 해석 질문

**Q1.** 이 분석은 AARRR 중 어느 단계에 해당하나요?
-> Activation 단계, 유로 구독자가 12월에 제품을 실제로 사용했는지를 확인하려고 했다.

**Q2.** 어떤 분석 기법을 사용했으며, 그 이유는 무엇인가요?
-> 세그멘테이션 분석을 사용, 요금제별로 구독을 나누어 제품의 활성률 차이를 비교했다.

**Q3.** 제품 활성률이 가장 낮은 요금제는 무엇인가요?
-> Pro, 19.37%로 가장 낮은 제품활성률을 보였다.

**Q4.** 해당 요금제에서 추가로 어떤 데이터를 확인하면 좋을까요?
-> 기능별 사용률, 오류 발생률, 기능 사용 시간, 고객 문의 등 추가 데이터 검토 필요

#### 제출 결과

- `plan_activation` 집계 코드와 결과
- AARRR 단계와 분석 기법
- 우선 점검할 요금제
- 추가 확인 데이터
- Q1~Q4 답변

```python
# 필수 2 코드를 작성하세요.
month_start = pd.Timestamp("2024-12-01")
month_end = pd.Timestamp("2024-12-31")

eligible_dec = subscriptions.loc[
    (subscriptions["is_trial"] == False)
    & (subscriptions["start_date"] <= month_end)
    & (
        subscriptions["end_date"].isna()
        | (subscriptions["end_date"] >= month_start)
    )
].copy()

active_subscription_ids = set(
    feature_usage.loc[
        (feature_usage["usage_date"] >= month_start)
        & (feature_usage["usage_date"] <= month_end)
        & (feature_usage["usage_count"] > 0),
        "subscription_id"
    ]
)

eligible_dec["is_product_active"] = eligible_dec["subscription_id"].isin(
    active_subscription_ids
)

plan_activation = eligible_dec.groupby("plan_tier").agg(
    eligible_paid_subscriptions=("subscription_id", "nunique"),
    active_paid_subscriptions=("is_product_active", "sum"),
    activation_rate=("is_product_active", "mean")
)
plan_activation = plan_activation.sort_values("activation_rate")

print(plan_activation.to_string(formatters={
    "activation_rate": lambda x: f"{x:.2%}"
}))

lowest_plan = plan_activation["activation_rate"].idxmin()
lowest_rate = plan_activation.loc[lowest_plan, "activation_rate"]

print(f"\n제품 활성률이 가장 낮은 요금제: {lowest_plan}")
print(f"해당 요금제의 제품 활성률: {lowest_rate:.2%}")
print("AARRR 단계: Activation")
print("분석 기법: 세그멘테이션 분석")
```

- `billing_activation` 집계 코드와 결과
- 우선 점검할 결제 주기
- 추가 확인 내용
- AARRR 단계와 분석 기법
- Q1~Q4 답변

```python
# 과제 1 코드를 작성하세요.
billing_activation = eligible_dec.groupby("billing_frequency").agg(
    eligible_paid_subscriptions=("subscription_id", "nunique"),
    active_paid_subscriptions=("is_product_active", "sum"),
    activation_rate=("is_product_active", "mean")
)

overall_activation_rate = eligible_dec["is_product_active"].mean()
billing_activation["gap_vs_overall"] = (
    billing_activation["activation_rate"] - overall_activation_rate
)

billing_activation = billing_activation.sort_values("activation_rate")

print(billing_activation.to_string(formatters={
    "activation_rate": lambda x: f"{x:.2%}",
    "gap_vs_overall": lambda x: f"{x:+.2%}p"}))

print(f"\n전체 제품 활성률: {overall_activation_rate:.2%}")

lowest_billing = billing_activation["activation_rate"].idxmin()
highest_billing = billing_activation["activation_rate"].idxmax()
lowest_rate = billing_activation.loc[lowest_billing, "activation_rate"]
highest_rate = billing_activation.loc[highest_billing, "activation_rate"]
lowest_eligible = billing_activation.loc[lowest_billing, "eligible_paid_subscriptions"]
lowest_active = billing_activation.loc[lowest_billing, "active_paid_subscriptions"]

print(f"\n제품 활성률이 가장 낮은 결제 주기: {lowest_billing} ({lowest_rate:.2%})")
print(f"- 이용 가능 유료 구독 수: {lowest_eligible:,}건 / 제품 활성 구독 수: {lowest_active:,}건")
print(f"- 전체 대비: {lowest_rate - overall_activation_rate:+.2%}p")
print(f"제품 활성률이 가장 높은 결제 주기: {highest_billing} ({highest_rate:.2%})")
print(f"결제 주기 간 격차: {highest_rate - lowest_rate:.2%}p")

print(f"\n우선 점검 대상: {lowest_billing}")
print("AARRR 단계: Activation / 분석 기법: 세그멘테이션 분석")

# Q3 근거: 결제 주기별로 요금제 구성이 치우쳐 있는지 확인
print("\n결제 주기 × 요금제 구성 비율")
cross = pd.crosstab(
    eligible_dec["billing_frequency"],
    eligible_dec["plan_tier"],
    normalize="index"
)
print(cross.to_string(float_format=lambda x: f"{x:.2%}"))
```
