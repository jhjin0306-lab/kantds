## 문제 2. 분석 가능한 고객 데이터로 정제

### 업무 상황

품질 진단 결과를 바탕으로 고객 데이터를 분석 가능한 상태로 정리해야 합니다.  
처리 과정에서 고객의 실제 구매 행동 정보가 불필요하게 손실되지 않도록 데이터 특성을 고려하세요.

### 요구사항

1. 원본 데이터를 보존한 상태에서 정제용 데이터를 생성하세요.
2. 분석에 맞지 않는 수치형·날짜형 자료형과 범주 표기를 정리하세요.
3. 결측치와 전체 행 기준 중복 데이터를 처리하세요.
4. 주요 수치형 컬럼의 이상치를 탐지하고 적절한 방법으로 처리하세요.
5. 처리 전후의 결측치, 중복값과 이상치 상태를 확인하세요.

### 힌트

- 변환할 수 없는 문자열은 결측치로 바꾼 뒤 일관되게 처리할 수 있습니다.
- 이상치는 무조건 삭제하기보다 실제 우수 고객의 행동일 가능성도 고려하세요.
- IQR은 이상치 후보를 확인하는 대표적인 방법입니다.

```python
# TODO: 원본을 보존하고 정제용 데이터 df_clean을 생성하세요.
df_clean = df.copy()
# TODO: 수치형·날짜형 자료형과 범주 표기를 정리하세요.
cols = ['Gender','Region','MembershipLevel','PreferredCategory','CouponUsed']
for col in cols:
    df_clean[col]=df_clean[col].astype(str).str.strip().str.title()
    df_clean[col] = df_clean[col].replace(["Nan", "nan", "NaN", "None", ""], np.nan)
    
df_clean['Age']= df_clean['Age'].astype(str).str.replace('세','',regex=False)
df_clean['Age'] = pd.to_numeric(df_clean['Age'], errors='coerce')

df_clean['MembershipLevel'] = df_clean['MembershipLevel'].replace({'Vip': 'VIP'})

df_clean['AveragePurchaseAmount']= df_clean['AveragePurchaseAmount'].astype(str).str.replace(',','',regex=False)
df_clean['AveragePurchaseAmount'] = pd.to_numeric(df_clean['AveragePurchaseAmount'], errors='coerce')

df_clean['LastPurchaseDate']= df_clean['LastPurchaseDate'].astype(str)
df_clean['LastPurchaseDate'] = df_clean['LastPurchaseDate'].str.replace('[./]','-',regex=True)
df_clean['LastPurchaseDate'] = pd.to_datetime(df_clean['LastPurchaseDate'], errors='coerce')

signup= df_clean['SignupDate'].astype(str).str.strip()
d1 = pd.to_datetime(signup, format="%Y-%m-%d", errors='coerce')
d2 = pd.to_datetime(signup, format= "%d-%m-%Y", errors='coerce')
df_clean['SignupDate'] = d1.fillna(d2)

df_clean['Gender']= df_clean['Gender'].replace({
    'M':'Male', 'f':'Female', 'F':'Female'
})
df_clean['CouponUsed']=df_clean['CouponUsed'].replace({
    'No':'N','no':'N', 'yes':'Y', 'Yes':'Y', 'YES':'Y'
})

df_clean['VisitCount']= pd.to_numeric(df_clean['VisitCount'], errors='coerce')
```

```python
# TODO: 결측치와 전체 행 기준 중복 데이터를 처리하세요.

df_clean = df_clean.drop_duplicates().reset_index(drop=True)

df_clean["Gender"] = df_clean["Gender"].fillna("Unknown")
print(df_clean["Gender"].value_counts(dropna=False))

df_other = df_clean[df_clean["Gender"] == "Other"]
print("Other 고객:", len(df_other), "명, 구매율", round(df_other["PurchaseStatus"].mean(), 3))

#나이 100이상// #수치형은 중앙값, 범주형은 최빈값
df_clean.loc[df_clean['Age'] > 100, 'Age'] = np.nan
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())
df_clean['SatisfactionScore'] = df_clean['SatisfactionScore'].fillna(df_clean['SatisfactionScore'].median())
df_clean['AveragePurchaseAmount'] = df_clean['AveragePurchaseAmount'].fillna(df_clean['AveragePurchaseAmount'].median())
df_clean['Region'] = df_clean['Region'].fillna(df_clean['Region'].mode()[0])
df_clean['PreferredCategory'] = df_clean['PreferredCategory'].fillna(df_clean['PreferredCategory'].mode()[0])
df_clean['MembershipLevel'] = df_clean['MembershipLevel'].fillna(df_clean['MembershipLevel'].mode()[0])
median_date = df_clean["LastPurchaseDate"].median()
df_clean["LastPurchaseDate"] = df_clean["LastPurchaseDate"].fillna(median_date)

# TODO: 처리 전후 결과를 확인하세요.
print(f"처리 전 행 수: {len(df)}")
print(f"처리 후 행 수: {len(df_clean)}")
print("중복:", df_clean.duplicated().sum(), "건")
print("결측:", df_clean.isnull().sum()) #결측치
# # 제거된 중복 행 수
# print(f"제거된 중복 행 수: {len(df) - len(df_clean)}")


# print("===== 최종 데이터 정보 =====")
# df_clean.info()

# print("\n===== 결측치 =====")
# print(df_clean.isna().sum())

# print("\n===== 중복 데이터 =====")
# print(df_clean.duplicated().sum())

# print("\n===== 데이터 타입 =====")
# print(df_clean.dtypes)
```

### 4-2. 결측치·이상치 처리 방법의 선택 근거

- 결측치 처리 방법:
-> Age의 결측은 중앙값로 대체했다.(빈칸 50)
-> Gender의 결측은 Unknown이라는 값으로 대체했다.(빈칸 26)
- 해당 방법을 선택한 이유:
-> Age : 130세 같이 큰 값으로 인해 평균이 올라가는데 중앙값은 이런 극단값의 영향을 평균보다 덜 받는다. 다만 빈칸 모두 중앙값으로 대체해 실제 나이를 알아낸 것은 아니다.
-> Gender : 여자 731명, 남자 696명으로 결측값을 어느 성별로 정하기엔 숫자의 차이가 크지 않았다. 그래서 Unknown(26)으로 따로 두었다. 또한 Other(47)는 성별 선택지 중의 하나로 만들었다.
- 이상치 처리 방법: IQR로 이상치 후보를 찾았다. 상한을 넘거나 하한보다 작으면 이상치로 간주했다. Age는 상한이 69.5세였고 100세 넘는 구매건은 잘못 입력된 값으로 봤다. 빈칸으로 처리 후 중앙값으로 대체했다.
- 이상치 처리 시 주의점: 상한보다 더 높은 값, 하한보다 더 낮은 값을 최빈값, 혹은 중앙값으로 대체한 것이라 이상치 처리 후 무조건 지우면 안된다.

- 파생변수 사용 시 한계: 
->20세와 29세가 똑같이 20대 이하가 되고,
   50만원 쓴 사람과 100만원 쓴 사람이 똑같이 High가 된다. 나이가 차이나도, 금액이 차이나도 같은 구간으로 묶여있으면 차이가 사라진다. 반대로 Q1 기준금액이 186,425원으로 186,200원과 186,500원은 300원 차이인데 등급이 갈린다.
->Age 결측을 중앙값인 39세로 채웠는데, 39세는 30대 구간의 끝이다.
   그래서 30대가 521명으로 가장 많아진 데에 영향이 있어 AgeGroup도 실제 연령대와 다를 수 있다.
-> 분위수 Q1, Q3는 데이터가 바뀌면 같이 달라져 기간별 등급을 바로 비교하기 어려울 수 있다.

```python
age_bins = [0, 29, 39, 49, 59, float("inf")]
age_labels = ["20대 이하", "30대", "40대", "50대", "60대 이상"]
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=age_bins,
    labels=age_labels,
    include_lowest=True
)
display(df_clean[["Age", "AgeGroup"]].head(10))
```

```python
purchase_q1 = df_clean["TotalPurchaseAmount"].quantile(0.25)
purchase_q3 = df_clean["TotalPurchaseAmount"].quantile(0.75)
print("Q1:", purchase_q1)
print("Q3:", purchase_q3)
df_clean["PurchaseGrade"] = "Medium"
low_condition = df_clean["TotalPurchaseAmount"] <= purchase_q1
df_clean.loc[low_condition, "PurchaseGrade"] = "Low"
high_condition = df_clean["TotalPurchaseAmount"] > purchase_q3
df_clean.loc[high_condition, "PurchaseGrade"] = "High"
df_clean["PurchaseGrade"].value_counts(dropna=False)
df_clean[["TotalPurchaseAmount", "PurchaseGrade"]].head(10)
```
