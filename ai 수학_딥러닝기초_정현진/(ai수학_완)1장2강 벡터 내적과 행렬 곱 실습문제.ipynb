### 문제 1-1. 내적으로 고객별 가중합 계산하기

#### 문제 설명

교안의 고객 활동 데이터와 가중치입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 2 |
| B | 5 | 1 |
| C | 2 | 4 |
| 가중치 | 2 | 4 |

입력과 가중치의 순서는 **방문 횟수 → 장바구니 추가 횟수**입니다. 이번 문제는 편향을 더하지 않고 내적만 계산합니다.

$$x\cdot w=x_1w_1+x_2w_2$$

- $x_1$, $x_2$: 고객의 첫 번째·두 번째 입력값
- $w_1$, $w_2$: 같은 위치의 특성에 적용할 가중치
- $\cdot$: 같은 위치끼리 곱한 값을 모두 더하는 내적 기호

#### 수식 읽는 순서

`x₁w₁`은 ‘첫 입력값 × 첫 가중치’입니다. 두 글자를 붙여 썼지만 곱셈을 뜻합니다. `x₂w₂`도 같은 방식입니다. `x·w`는 이 곱셈 결과들을 모두 더한 값입니다.

‘대응하는 원소’는 **같은 순서에 있는 숫자끼리**라는 뜻입니다. 방문 횟수에는 방문 가중치를, 장바구니 횟수에는 장바구니 가중치를 곱하세요.

````python
# 문제 1-1. 내적으로 고객별 가중합 계산하기
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 설명을 참고하세요.

w = np.array([2,4]) 
x_a = np.array([3,2])
x_b = np.array([5,1])
x_c = np.array([2,4])

score_a = np.dot(x_a, w)
score_b = np.dot(x_b, w)
score_c = np.dot(x_c, w)
print("A 점수:", score_a)
# 고객 A는 3X2=6, 2X4=8, 총합 = 14
print("B 점수:", score_b)
# 고객 B는 5X2=10, 1X4=4, 총합 = 14
print("C 점수:", score_c) 
# 고객 C는 2X2=4, 4X4=16, 총합 = 20

a_w = x_a @ w
print('\n A 점수(@) :',a_w)
print(np.allclose(a_w, score_a))
````

### 문제 1-2. 내적의 부호와 코사인 유사도 이해하기

#### 문제 설명

교안의 간단한 방향 예시를 사용합니다. 아래의 모든 벡터는 0벡터가 아닙니다.

| 구분 | 벡터 |
|---|---|
| 기준 벡터 `a` | [1, 0] |
| 같은 방향 `b_same` | [2, 0] |
| 수직 방향 `b_perp` | [0, 2] |
| 반대 방향 `b_opposite` | [−2, 0] |

추가로 교안의 `u = [3, 4]`, `v = [6, 8]`의 코사인 유사도를 계산합니다.

$$\mathrm{cosine\ similarity}=\frac{u\cdot v}{\|u\|_2\times\|v\|_2}$$

분자는 두 벡터의 내적, 분모는 두 벡터의 L2 Norm을 곱한 값입니다.

| 결과 | 방향 관계 |
|---|---|
| 코사인 유사도 1 | 같은 방향 |
| 코사인 유사도 0 | 서로 수직 |
| 코사인 유사도 −1 | 반대 방향 |

> 교안의 코사인 유사도 목록 중 ‘반대 방향: 1’은 ‘반대 방향: −1’의 부호 오타이므로 바로잡아 적용했습니다.

#### 먼저 이해하기

- 벡터를 원점에서 출발하는 화살표로 생각하면, ‘방향’은 화살표가 향하는 쪽입니다.
- `[1,0]`과 `[2,0]`은 같은 쪽, `[1,0]`과 `[0,2]`는 직각, `[1,0]`과 `[−2,0]`은 서로 반대쪽을 향합니다.
- 90°는 직각, 180°는 완전히 반대 방향입니다. 각도를 구하는 코드는 작성하지 않고 표와 내적의 부호를 연결합니다.
- **분자**는 분수의 위쪽, **분모**는 아래쪽입니다. 코사인 유사도 식은 ‘내적 ÷ (첫 길이 × 둘째 길이)’로 읽습니다.
- L2 길이는 이전 강의처럼 ‘각 원소 제곱 → 더하기 → 제곱근’으로 구합니다. 예를 들어 √25는 5×5=25이므로 5입니다.

1~2번은 방향별 내적, 3~4번은 u·v의 코사인 유사도입니다. 두 부분을 차례로 진행하세요.


- X의 행 순서: 고객 A, B, C
- X의 열 순서: 방문 횟수, 장바구니 추가 횟수

행렬 두 개의 곱은 다음 규칙을 따릅니다.

$$(m,n)\ @\ (n,p)\ \longrightarrow\ (m,p)$$

m은 고객 수, n은 입력 특성 수, p는 출력 수로 읽을 수 있습니다. 가운데 크기가 같아야 계산할 수 있습니다.

#### Shape 규칙을 말로 읽기

두 행렬의 Shape는 `(행 수, 열 수)`입니다. `(m,n) @ (n,p)`에서 먼저 **왼쪽 열 수와 오른쪽 행 수**를 비교합니다. 이 둘이 같으면 한 행과 한 열의 숫자를 빠짐없이 짝지어 내적할 수 있습니다.

결과의 행 수는 왼쪽 행렬에서, 결과의 열 수는 오른쪽 행렬에서 가져옵니다. 이 문제에서는 **고객 수 × 출력 점수 종류 수**입니다.

1~2번은 고객별 한 점수, 3~4번은 고객별 두 점수, 5번은 Shape를 읽는 연습입니다. 5번은 실제 배열을 만들지 않고 숫자의 개수만 비교하세요.

````python
## 코드로 회귀 역전파 확인하기

x, y = 2.0, 10.0
w, b = 3.0, 1.0
learning_rate = 0.01

#1. 순전파의 손실계산
y_pred = w * x + b # <----------------레이어 1
loss = (y- y_pred) **2

#2. 손실에서 예측값으로 (역전파)
dL_dy_pred = 2 * (y_pred - y)

#3. 예측값에서 각 파라미터로 (역전파)
dL_dw = dL_dy_pred * x
dL_db = dL_dy_pred * 1.0

print("예측과 손실", y_pred, loss)
print("기울기", dL_dy_pred, dL_dw, dL_db)
print("업데이트 전", w, b)

#4. 모든 기울기를 구한 다음 업데이트 진행
new_w = w - learning_rate * dL_dw
new_b = b - learning_rate * dL_db
new_pred = new_w * x + new_b
new_loss = (y - new_pred)**2
print("업데이트 후", new_w, new_b)
print("새 손실",round(new_loss,2))
````

````python
import numpy as np

### 코드로 은닉층과 역전파 확인하기
x, y = 2.0, 1.0
w1, b1 = 1.0, 0.0
w2, b2 = 0.5, 0.0

#1. 순전파
a = w1 * x + b1
````
````python
h = max(0,0,a)  #<----------------여기까지 레이어 1, 뉴런 1개의 계산, 은닉층의 뉴런 하나가 입력을 받아 출력값 h를 받는다.
                #<------뉴런이 3개? 같은 입력 x를 받지만 뉴런은 자기 가중치와 자기 편향으로 따로 계산
                # 은닉층 하나에 뉴런 3개 h1 = max(0.0, w1 * x + b1) h2 = max(0.0, w2 * x+b2)...
                # 따라서 은닉층의 출력이 [h1, h2, h3] 앞에 뉴런을 차례대로 계산하는게 하나라 같은 층에서 나란히 계산하기 때문 
z = w2 * h + b2
p = 1.0 / (1.0+np.exp(-z)) # <--------------여기까지 레이어2
loss = np.logaddexp(0.0,z) - y * z # <-------------------손실은 레이어 아님

#2. 출력층의 역전파
dL_dz = p - y # <---------------------------------레이어 x, 출력층의 기울기 계산
dL_dw2 = dL_dz * h
dL_db2 = dL_dz

#ReLU를 거쳐 은닉층으로 전달 <--------------------------------레이어 x, 은닉층 기울기 계산
dL_dh = dL_dz * w2
relu_grad = 1.0 if a > 0 else 0.0
dL_da = dL_dh * relu_grad
dL_dw1 = dL_da * x
dL_db1 = dL_da
print("중간값", a, h, z)
print("확률과 손실", round(p, 4), round(loss,4))
print("첫 층", round(dL_dw1, 4), round(dL_db1,4))
print("출력층", round(dL_dw2,4), round(dL_db2,4))
````
