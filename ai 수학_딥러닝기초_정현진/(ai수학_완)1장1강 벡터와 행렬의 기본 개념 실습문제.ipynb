### 문제 1-2. 입력·정답·가중치·편향 구분과 점수 계산

#### 문제 설명

문제 1-1과 같은 고객 데이터에 실제 구매 여부를 추가했습니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 | 실제 구매 여부 |
|---|---:|---:|---:|
| A | 5 | 2 | 1 |
| B | 3 | 1 | 0 |
| C | 8 | 4 | 1 |

구매한 경우는 1, 구매하지 않은 경우는 0입니다. 방문 횟수의 가중치는 2, 장바구니 추가 횟수의 가중치는 3, 편향은 1로 정합니다.

교안의 계산식을 그대로 사용합니다.

$$\mathrm{score}=w_1x_1+w_2x_2+b$$

| 기호 | 의미 |
|---|---|
| $x_1$, $x_2$ | 방문 횟수, 장바구니 추가 횟수 |
| $w_1$, $w_2$ | 각 특성에 곱하는 가중치 |
| $b$ | 마지막에 더하는 편향 |
| score | 가중합으로 계산한 출력 점수 |

또한 교안의 두 출력용 가중치 표를 행렬로 표현해 보세요.

| 입력 특성 | 구매 관심도 가중치 | 고객 활동 가중치 |
|---|---:|---:|
| 방문 횟수 | 2 | 1 |
| 장바구니 추가 횟수 | 3 | 4 |

#### 수식을 말로 읽기

`w₁x₁`은 **첫 가중치 × 첫 입력값**입니다. 두 글자가 붙어 있어도 곱셈을 뜻합니다. 아래첨자 1과 2는 첫 번째·두 번째를 구분하는 번호입니다.

전체 식은 ‘방문 횟수에 해당 가중치를 곱한다 → 장바구니 추가 횟수에도 해당 가중치를 곱한다 → 두 결과와 편향을 더한다’는 뜻입니다. **가중합**은 이렇게 ‘각각 곱한 뒤 합한 값’입니다. **편향**은 그 합에 마지막으로 더하는 값입니다.

#### 요구사항

### 문제 2-1. 음수가 포함된 벡터의 L1 Norm과 L2 Norm

#### 문제 설명

교안의 벡터 `v = [3, -4]`를 사용합니다. 이 벡터는 노름 계산을 연습하기 위한 숫자 예시입니다.

노름은 벡터의 크기를 하나의 숫자로 표현합니다. 두 원소에 대한 공식은 다음과 같습니다.

$$\|v\|_1=|v_1|+|v_2|$$
$$\|v\|_2=\sqrt{v_1^2+v_2^2}$$

- $v_1$, $v_2$: 벡터의 첫 번째와 두 번째 원소
- $|a|$: a의 절댓값. 예를 들어 −4의 절댓값은 4입니다.
- $a^2$: a를 자기 자신과 곱한 값
- $\sqrt{a}$: 제곱하여 a가 되는 수 중 음수가 아닌 값. 예를 들어 $\sqrt{25}=5$입니다.
- $\|v\|_1$, $\|v\|_2$: 각각 벡터 v의 L1 Norm, L2 Norm

#### 수식을 말로 읽기

- `||v||₁`은 ‘벡터 v의 엘원 노름’, `||v||₂`는 ‘벡터 v의 엘투 노름’으로 읽습니다. 아래의 1과 2는 **계산 방법의 종류**입니다.
- 절댓값은 숫자가 0에서 얼마나 떨어져 있는지 나타냅니다. 양수는 그대로이고 음수는 크기만 남깁니다.
- 제곱은 같은 수를 두 번 곱하는 것입니다. 음수 전체를 제곱할 때는 `(-4)²`처럼 괄호로 묶어 읽으세요.
- 제곱근 기호 √는 ‘어떤 음수가 아닌 수를 두 번 곱해야 이 값이 되는가’를 묻습니다.

#### 요구사항

2. `np.linalg.norm(v, ord=1)`로 L1 Norm을 계산하고 출력하세요.
3. `np.linalg.norm(v, ord=2)`로 L2 Norm을 계산하고 출력하세요.
4. 답변 작성란에 두 공식의 대입·계산 과정을 적고 코드 결과와 비교하세요.

#### 작성 도움말

`ord=1`은 절댓값의 합, `ord=2`는 제곱합의 제곱근을 계산하도록 선택합니다.

#### 해석 질문

- v와 두 노름 출력
- 공식에 값을 대입한 계산 과정 및 Q1~Q4 답변

```python
# 문제 2-1. 음수가 포함된 벡터의 L1 Norm과 L2 Norm
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 작성 도움말을 참고하세요.

v = np.array([3,-4]) # 첫번째 원소는 3, 두번째 원소는 -4인 벡터 v를 생성 
l1_norm = np.linalg.norm(v, ord=1) # v의 노름을 구하는 L1 Norm 계산
# L1 계산식은 절대값 -> 덧셈
# |3| + |-4| = 3 + 4 = 7

l2_norm = np.linalg.norm(v, ord=2) # v의 노름을 구하는 L2 Norm 계산
# L2 계산식은 제곱합의 제곱근
# sqrt(3^2 + (-4)^2) = sqrt(9 + 16) = sqrt(25) = 5

print("벡터 v:", v)
print("L1 Norm:", l1_norm)
print("L2 Norm:", l2_norm)      
```

- 작성:

* |3| + |-4| = 3 + 4 = 7

* sqrt(3^2 + (-4)^2) = sqrt(9 + 16) = sqrt(25) = 5

- 3 + 4 = 7 로 계산이 됨, 절대값 계산 후 더하는 순서로 진행하는 것이 특징

**Q2. L2 Norm은 얼마인가요? 각 원소의 제곱, 합, 제곱근 순서로 계산 과정을 적으세요.**

- 답변: 5 // 첫 원소 3의 제곱 = 3 X 3 = 9, 두번쨰 원소는 (-4)의 제곱 = (-4) X (-4) = 16,
- 9 + 16 = 25고, 25에다가 루트를 반영하면 5
- 제곱근을 반영하여 끝까지 구하는 것이 특징 

**Q3. 원소를 그냥 더한 `3 + (-4)`와 L1 Norm이 다른 이유는 무엇인가요?**

- 답변: 단순 덧셈으로 계산을 하면 +3 -4 = -1인데 
- L1 Norm은 절대값을 구한 뒤 덧셈을 진행하기 때문에 7이다.
- 즉, 방향성이 다른 원소가 있어도 크기 계산에서 서로 상쇄하지 않는다.

**Q4. 원소가 2개라는 사실과 L2 Norm의 계산 결과는 같은 뜻인가요?**

### 문제 2-2. 두 고객 사이의 거리와 벡터 정규화

#### 문제 설명

교안의 연습문제와 같은 두 고객의 활동 데이터입니다.

| 고객 | 방문 횟수 | 장바구니 추가 횟수 |
|---|---:|---:|
| A | 3 | 4 |
| B | 6 | 8 |

원소 순서는 방문 횟수 → 장바구니 추가 횟수입니다.

두 고객 사이의 거리는 **B 벡터 − A 벡터**에 노름을 적용합니다.

$$d_1(A,B)=\|x_B-x_A\|_1,\qquad d_2(A,B)=\|x_B-x_A\|_2$$

길이가 1인 벡터를 만들 때는 각 벡터를 **자신의 L2 Norm**으로 나눕니다.

$$x_{\mathrm{normalized}}=\frac{x}{\|x\|_2}$$

이 방법은 L2 Norm이 0이 아닌 벡터에 적용합니다.

#### 수식을 말로 읽기

`x_B−x_A`는 **같은 위치의 특성끼리 뺀다**는 뜻입니다. 방문 횟수끼리, 장바구니 추가 횟수끼리 뺍니다. `d₁(A,B)`와 `d₂(A,B)`는 A와 B 사이의 거리를 두 방법으로 계산한 값입니다.

정규화 식의 분수는 ‘위에 있는 벡터의 **모든 원소**를 아래의 L2 Norm 값으로 나눈다’고 읽습니다. 길이가 1이라는 뜻은 **L2 Norm이 1**이라는 뜻입니다.

**1~3번은 거리**, **4~5번은 원래 길이를 구한 뒤 정규화**, **6~7번은 결과 확인**입니다. 한 묶음씩 진행하세요.

**Q5.** `[0, 0]`을 같은 방법으로 정규화하면 안 되는 이유는 무엇인가요?

#### 제출 결과

- 차이 벡터, 맨해튼·유클리드 거리
- 정규화 전후 L2 Norm, 정규화 벡터와 비교 결과
- Q1~Q5 답변

```python
# 문제 2-2. 두 고객 사이의 거리와 벡터 정규화
# 요구사항을 한 단계씩 작성하고 실행하세요.
# 계산 방법과 함수 이름은 위의 작성 도움말을 참고하세요.

x_a = np.array([3,4])  
x_b = np.array([6,8])   

diff = x_b - x_a  # 두 벡터의 차이 계산
## 방문횟수 = 6-3 = 3
## 장바구니 추가 횟수 = 8-4 = 4

manhattan_distance = np.linalg.norm(diff, ord=1)  # L1 Norm 계산
euclidean_distance = np.linalg.norm(diff, ord=2)  # L2 Norm 계산
print("차이 벡터:", diff)
print("맨하탄 거리:", manhattan_distance)
print("유클리드 거리:", euclidean_distance)

a_12 = np.linalg.norm(x_a, ord=2)  # x_a의 L2 Norm 계산
b_12 = np.linalg.norm(x_b, ord=2)  # x_b의 L2 Norm 계산
print("A 정규화 전 L2:", a_12)
print("B 정규화 전 L2:", b_12)

a_normalized = x_a / a_12  # x_a를 L2 Norm으로 나누어 정규화
## A 정규화 : 3/5 = 0.6, 4/5 = 0.8
b_normalized = x_b / b_12  # x_b를 L2 Norm으로 나누어 정규화
## B 정규화 : 6/10 = 0.6, 8/10 = 0.8
## 1. 0.6의 제곱 + 0.8의 제곱 = 0.36 + 0.64 = 1.0 (루트계산 생략)

print("A 정규화 후:", a_normalized)
print("B 정규화 후:", b_normalized) 
print("A 정규화 후 L2:", np.linalg.norm(a_normalized, ord=2))   
print("B 정규화 후 L2:", np.linalg.norm(b_normalized, ord=2))   
print("정규화 결과가 같은가?:", np.allclose(a_normalized, b_normalized))
```

**Q2. A 벡터의 L2 Norm과 A·B 사이의 유클리드 거리는 각각 무엇을 측정하나요?**

- L2 Norm: 각 원소를 제곱하고, 더한 뒤, 제곱근을 구합니다.
